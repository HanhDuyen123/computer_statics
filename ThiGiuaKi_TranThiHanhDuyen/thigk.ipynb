```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
data=pd.read_csv('penguins_data.csv')
data.head(10)
```

```python
d0=data[data['species']=='Adelie']
d1=data[data['species']=='Chinstrap']
d2=data[data['species']=='Gentoo']
index=['Count']
df=pd.DataFrame({
    'index': index,
    'Adelie': d0['island'].count(),
    'Chinstrap': d1['island'].count(),
    'Gentoo': d2['island'].count()
})
print(df)
```

```python
data_island = data.groupby(['island', 'species']).size().unstack(fill_value=0)
df1=pd.DataFrame({
    'Chinstrap': data_island.get('Chinstrap', [0,0,0],),
    'Adelie': data_island.get('Adelie', [0,0,0],),
    'Gentoo': data_island.get('Gentoo', [0,0,0],)
}, index=['Biscoe', 'Dream', 'Torgersen'])
print(df1)
```

```python
d0=data[data['species']=='Adelie']
d1=data[data['species']=='Chinstrap']
d2=data[data['species']=='Gentoo']
index=['Mean', 'Median', 'Std']
x=[d0['bill_length_mm'].mean(),d0['bill_length_mm'].median(), d0['bill_length_mm'].std()]
y=[d1['bill_length_mm'].mean(),d1['bill_length_mm'].median(), d1['bill_length_mm'].std()]
z=[d2['bill_length_mm'].mean(),d2['bill_length_mm'].median(), d2['bill_length_mm'].std()]
df=pd.DataFrame({
    'index': index,
    'Adelie': x,
    'Chinstrap': y,
    'Gentoo': z
})
print(df)
```

```python
data=pd.DataFrame({
    'Biscoe': [0, 124, 44],
    'Dream': [68, 0, 56],
    'Torgersen': [0, 0, 52]
}, index=['Chinstrap', 'Gentoo', 'Adelie'])

data.T.plot(kind='bar', stacked=True, color=['lightskyblue', 'lightgreen', 'pink'])
plt.title('Biểu đồ phân phối của ba loài chim cánh cụt')
plt.xlabel('Đảo')
plt.ylabel('Số lượng loài')
plt.show()
```
