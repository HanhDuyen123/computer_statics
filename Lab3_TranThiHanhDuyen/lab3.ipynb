```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
data=pd.read_csv('babies.txt',delim_whitespace=True)

d0=data[data['smoke']==0]
d1=data[data['smoke']==1]

index=['Count', 'Min', 'Max', 'Mean', 'std', 'Var', 'Median', '0%', '25%', '50%', '75%', 'IQR', 'Skewness', 'Kurtosis']
x=[
    d1['bwt'].count(), d1['bwt'].min(), d1['bwt'].max(),
    d1['bwt'].mean(), d1['bwt'].std(), d1['bwt'].var(), d1['bwt'].median(),
    d1['bwt'].quantile(0), d1['bwt'].quantile(0.25), d1['bwt'].quantile(0.5), d1['bwt'].quantile(0.75),
    d1['bwt'].quantile(0.75)-d1['bwt'].quantile(0.25),
    d1['bwt'].skew(), d1['bwt'].kurtosis()
]
y=[
    d0['bwt'].count(), d0['bwt'].min(), d0['bwt'].max(),
    d0['bwt'].mean(), d0['bwt'].std(), d0['bwt'].var(), d0['bwt'].median(),
    d0['bwt'].quantile(0), d0['bwt'].quantile(0.25), d0['bwt'].quantile(0.5), d0['bwt'].quantile(0.75),
    d0['bwt'].quantile(0.75)-d0['bwt'].quantile(0.25),
    d0['bwt'].skew(), d0['bwt'].kurtosis()
]

df=pd.DataFrame({
    'index': index,
    'TH1: Nguoi me co hut thuoc': x,
    "TH2: Nguoi me khong hut thuoc": y,
})
print(df)
```
