```python
small_pop=np.array([151,158,162,173,180])
print(small_pop)
mean_small_pop=np.mean(small_pop)
print('Chieu cao trung binh cua quan the: {}'.format(mean_small_pop))
```

```python
np.random.seed(24)
samp1=np.random.choice(small_pop, size=4, replace=False)
mean_samp1=np.mean(samp1)
print('Mẫu: ',samp1)
print('Chiều cao trung bình mẫu: {}'.format(mean_samp1))
print('Sai số: {}'.format(abs(mean_small_pop-mean_samp1)))
```
